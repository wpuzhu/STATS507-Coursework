```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

# Dataset Loading and Preprocessing

## Data Cleaning

```python
t1_data = pd.read_excel("../multi_omics/T1.xlsx")
t2_data = pd.read_excel("../multi_omics/T2.xlsx")
adc_data = pd.read_excel("../multi_omics/ADC.xlsx")
post_survey = pd.read_excel("../multi_omics/post_survey.xlsx")
```

```python
# Step 1: Build aligned imaging dataset
merged = t1_data.merge(
    t2_data,
    on=["case", "center"],
    suffixes=("_t1", "_t2")
)
merged = merged.merge(adc_data, on=["case", "center"])

# Rename ADC feature columns to avoid confusion
adc_cols = [
    col for col in merged.columns
    if col not in ["case", "center"]
    and not col.endswith("_t1")
    and not col.endswith("_t2")
]

merged = merged.rename(
    columns={col: f"{col}_adc" for col in adc_cols if col != "Label"}
)

# Drop rows with ANY NA across all imaging modalities
merged = merged.dropna().reset_index(drop=True)

print("After dropping rows with NA across all modalities:", merged.shape)

# Step 2: Extract labels
t1_label = merged["Label_t1"].values
t2_label = merged["Label_t2"].values
adc_label = merged["Label"].values

assert np.all(t1_label == t2_label)
assert np.all(t1_label == adc_label)

# Step 3: Extract mri features
t1_features = [col for col in merged.columns if col.endswith("_t1") and col != "Label_t1"]
t2_features = [col for col in merged.columns if col.endswith("_t2") and col != "Label_t2"]
adc_features = [col for col in merged.columns if col.endswith("_adc")]

t1_data_clean = merged[t1_features].copy()
t2_data_clean = merged[t2_features].copy()
adc_data_clean = merged[adc_features].copy()

print("Final aligned imaging shapes:")
print("T1:", t1_data_clean.shape)
print("T2:", t2_data_clean.shape)
print("ADC:", adc_data_clean.shape)

# Step 4: Keep the same patients in post_survey
if {"case", "center"}.issubset(post_survey.columns):
    patient_keys = merged[["case", "center"]].drop_duplicates()

    post_survey_aligned = patient_keys.merge(
        post_survey,
        on=["case", "center"],
        how="left"   # keep all imaging patients, even if post_survey is missing
    )

print("Aligned post_survey shape:", post_survey_aligned.shape)
print("Missing rows in aligned post_survey:",
      post_survey_aligned.isna().all(axis=1).sum())
post_survey_aligned.to_csv("../multi_omics/post_survey_aligned.csv", index=False)
print("Saved to ../multi_omics/post_survey_aligned.csv")
```

## Train Test Split 

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_t1 = t1_data_clean.values if isinstance(t1_data_clean, pd.DataFrame) else t1_data_clean
X_t2 = t2_data_clean.values if isinstance(t2_data_clean, pd.DataFrame) else t2_data_clean
X_adc = adc_data_clean.values if isinstance(adc_data_clean, pd.DataFrame) else adc_data_clean

y = np.array(t1_label)

t1_names = np.array(t1_features)
t2_names = np.array(t2_features)
adc_names = np.array(adc_features)

# 1. Train/Test split (shared across modalities)
indices = np.arange(len(y))

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.2,
    random_state=42,
    stratify=y
)

# 2. Train/Test split to all modalities 
def split_data(X, train_idx, test_idx):
    return X[train_idx], X[test_idx]

t1_X_train, t1_X_test = split_data(X_t1, train_idx, test_idx)
t2_X_train, t2_X_test = split_data(X_t2, train_idx, test_idx)
adc_X_train, adc_X_test = split_data(X_adc, train_idx, test_idx)

y_train = y[train_idx]
y_test = y[test_idx]

# 3. Standardization (fit on train only to avoid data leakage)
def standardize(train, test):
    scaler = StandardScaler()
    train_std = scaler.fit_transform(train)
    test_std = scaler.transform(test)
    return train_std, test_std

t1_train_std, t1_test_std = standardize(t1_X_train, t1_X_test)
t2_train_std, t2_test_std = standardize(t2_X_train, t2_X_test)
adc_train_std, adc_test_std = standardize(adc_X_train, adc_X_test)

# 4. Save standardized datasets WITH label
def save_with_label(train, test, y_train, y_test, feature_names, prefix):
    
    # TRAIN
    train_df = pd.DataFrame(train, columns=feature_names)
    train_df["label"] = y_train   
    
    # TEST
    test_df = pd.DataFrame(test, columns=feature_names)
    test_df["label"] = y_test
    
    # SAVE
    train_df.to_csv(f"../multi_omics/{prefix}_train_std.csv", index=False)
    test_df.to_csv(f"../multi_omics/{prefix}_test_std.csv", index=False)

save_with_label(t1_train_std, t1_test_std, y_train, y_test, t1_names, "T1")
save_with_label(t2_train_std, t2_test_std, y_train, y_test, t2_names, "T2")
save_with_label(adc_train_std, adc_test_std, y_train, y_test, adc_names, "ADC")

# 5. Check shapes
print("T1:", t1_train_std.shape, t1_test_std.shape)
print("T2:", t2_train_std.shape, t2_test_std.shape)
print("ADC:", adc_train_std.shape, adc_test_std.shape)
```

## Feature Selection

```python
from sklearn.feature_selection import SelectKBest, f_classif

# load standardized train/test datasets
t1_train_std = pd.read_csv("../multi_omics/T1_train_std.csv")
t1_test_std = pd.read_csv("../multi_omics/T1_test_std.csv")

t2_train_std = pd.read_csv("../multi_omics/T2_train_std.csv")
t2_test_std = pd.read_csv("../multi_omics/T2_test_std.csv")

adc_train_std = pd.read_csv("../multi_omics/ADC_train_std.csv")
adc_test_std = pd.read_csv("../multi_omics/ADC_test_std.csv")


def univariate_selection_fit(X_train, y_train, k=50):
    k = min(k, X_train.shape[1])
    selector = SelectKBest(score_func=f_classif, k=k)
    X_train_new = selector.fit_transform(X_train, y_train)
    mask = selector.get_support()
    return X_train_new, mask


def correlation_filter_fit(X_train, threshold=0.9):
    corr = np.corrcoef(X_train, rowvar=False)
    n = corr.shape[0]
    keep = np.ones(n, dtype=bool)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(corr[i, j]) > threshold and keep[j]:
                keep[j] = False

    X_train_new = X_train[:, keep]
    return X_train_new, keep


def feature_selection_pipeline(train_df, test_df, k=50, corr_thresh=0.9, verbose=True):
    y_train = train_df["label"].values
    y_test = test_df["label"].values

    X_train = train_df.drop(columns=["label"]).values
    X_test = test_df.drop(columns=["label"]).values

    feature_names = np.array(train_df.drop(columns=["label"]).columns)

    if verbose:
        print(f"Original number of features: {X_train.shape[1]}")

    X_train_sel1, mask1 = univariate_selection_fit(X_train, y_train, k=k)
    X_test_sel1 = X_test[:, mask1]
    names1 = feature_names[mask1]

    if verbose:
        print(f"After SelectKBest: {X_train_sel1.shape[1]} features")

    X_train_sel2, mask2 = correlation_filter_fit(X_train_sel1, threshold=corr_thresh)
    X_test_sel2 = X_test_sel1[:, mask2]
    names2 = names1[mask2]

    if verbose:
        print(f"After correlation filtering: {X_train_sel2.shape[1]} features")

    train_out = pd.DataFrame(X_train_sel2, columns=names2)
    train_out["label"] = y_train

    test_out = pd.DataFrame(X_test_sel2, columns=names2)
    test_out["label"] = y_test

    return train_out, test_out, names2


# Apply feature selectionto each modality
print("T1")
t1_train_sel, t1_test_sel, t1_selected_names = feature_selection_pipeline(
    t1_train_std, t1_test_std, k=50, corr_thresh=0.8
)

print("\nT2")
t2_train_sel, t2_test_sel, t2_selected_names = feature_selection_pipeline(
    t2_train_std, t2_test_std, k=50, corr_thresh=0.8
)

print("\nADC")
adc_train_sel, adc_test_sel, adc_selected_names = feature_selection_pipeline(
    adc_train_std, adc_test_std, k=50, corr_thresh=0.8
)

# Save all selected datasets
t1_train_sel.to_csv("../multi_omics/T1_train_selected.csv", index=False)
t1_test_sel.to_csv("../multi_omics/T1_test_selected.csv", index=False)

t2_train_sel.to_csv("../multi_omics/T2_train_selected.csv", index=False)
t2_test_sel.to_csv("../multi_omics/T2_test_selected.csv", index=False)

adc_train_sel.to_csv("../multi_omics/ADC_train_selected.csv", index=False)
adc_test_sel.to_csv("../multi_omics/ADC_test_selected.csv", index=False)

train_dict = {
    "T1" : (t1_train_sel.iloc[:, :-1], t1_train_sel.iloc[:, -1]),
    "T2" : (t2_train_sel.iloc[:, :-1], t2_train_sel.iloc[:, -1]),
    "ADC" : (adc_train_sel.iloc[:, :-1], adc_train_sel.iloc[:, -1])
}

print("\nSaved shapes:")
print("T1:", t1_train_sel.shape, t1_test_sel.shape)
print("T2:", t2_train_sel.shape, t2_test_sel.shape)
print("ADC:", adc_train_sel.shape, adc_test_sel.shape)
```

# Explanatory Data Analysis

## Hierarchical Cluster Heatmap

```python
def plot_clustered_corr_heatmap(data, modality, threshold=0.7, figsize=(14, 14), annotate=False):
    df = pd.DataFrame(data)

    if "label" in df.columns:
        df = df.drop(columns=["label"])

    corr = df.corr()

    strong_mask = (np.abs(corr) >= threshold)
    np.fill_diagonal(strong_mask.values, False)
    keep_features = strong_mask.any(axis=0)

    corr_sub = corr.loc[keep_features, keep_features]

    if corr_sub.shape[0] < 2:
        print(f"{modality}: not enough strongly correlated features to cluster.")
        return

    g = sns.clustermap(
        corr_sub,
        cmap="coolwarm",
        center=0,
        vmin=-1,
        vmax=1,
        linewidths=0.3,
        figsize=figsize,
        annot=annotate,
        fmt=".2f",
        annot_kws={"size": 7},
        cbar_kws={"shrink": 0.2},
        cbar_pos=(0.02, 0.85, 0.03, 0.15)
    )

    g.fig.suptitle(
        f"{modality} Hierarchical Clustered Correlation Heatmap (|corr| ≥ {threshold})",
        y=1.02,
        fontsize=16
    )

    plt.setp(g.ax_heatmap.get_xticklabels(), rotation=60, ha="right", fontsize=9)
    plt.setp(g.ax_heatmap.get_yticklabels(), rotation=0, fontsize=9)

    plt.show()

for modality, (data, label) in train_dict.items():
    plot_clustered_corr_heatmap(data, modality, threshold=0.7, annotate=False)
```

## Local Neighborhood Structure -- 2D T-SNE

```python
from sklearn.manifold import TSNE

def run_tsne(X, perplexity=30, random_state=42):
    X = np.asarray(X)
    tsne = TSNE(
        n_components=2,
        perplexity=perplexity,
        random_state=random_state,
        init="pca",
        learning_rate="auto",
        max_iter=1000
    )
    return tsne.fit_transform(X)

t1_tsne = run_tsne(train_dict["T1"][0])
t2_tsne = run_tsne(train_dict["T2"][0])
adc_tsne = run_tsne(train_dict["ADC"][0])

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

datasets = [
    ("T1", t1_tsne, train_dict["T1"][1]),
    ("T2", t2_tsne, train_dict["T2"][1]),
    ("ADC", adc_tsne, train_dict["ADC"][1]),
]

for ax, (name, emb, y) in zip(axes, datasets):
    y = np.asarray(y).ravel()

    for label in np.unique(y):
        mask = y == label
        ax.scatter(
            emb[mask, 0],
            emb[mask, 1],
            alpha=0.7,
            s=35,
            label=f"Label {label}"
        )

    ax.set_title(name)
    ax.set_xlabel("t-SNE 1")
    ax.set_ylabel("t-SNE 2")
    ax.grid(alpha=0.3)

handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="upper center", ncol=2)
plt.tight_layout(rect=[0, 0, 1, 0.92])
plt.show()
```
